```python
# Calculate the percentage of missing values for each column
missing_percentage = df.isnull().mean() * 100

# Print the percentage of missing values
print(missing_percentage)
```

```python
column_means = df.mean()

# Fill missing values with the mean of their respective column
df_filled = df.fillna(column_means)
```

```python
df_filled.isnull().sum()
```

```python
# Convert categorical variable into dummy/indicator variables
df_encoded = pd.get_dummies(df_filled, prefix='Category')

print(df_encoded)
```

```python
df_encoded.head()
```

```python
df_encoded.isnull().sum()
```

```python
y = df_encoded['Area Planted (acres)']
X = df_encoded.drop('Area Planted (acres)', axis = 1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsRegressor
knn_regressor = KNeighborsRegressor(n_neighbors=5)
knn_regressor.fit(X_train_scaled, y_train)
```

```python
from sklearn.metrics import mean_squared_error
y_pred = knn_regressor.predict(X_test_scaled)

# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
```

```python
from sklearn.metrics import r2_score
r_squared = r2_score(y_test, y_pred)
print(f"R-squared: {r_squared}")
```
