# Importing Libraries

```python
import numpy as np
import heapq
import time
```

```python
def main():
    # Constants
    N = 34279  # Total number of pages
    K = 20     # Number of top pages to extract
    min_weight = -80000
    max_weight = 50000

    # Generate random page weights
    np.random.seed(0)  # Seed for reproducibility
    page_weights = np.random.uniform(min_weight, max_weight, N)

    # Convert numpy array to list for compatibility with heapq
    page_weights_list = list(page_weights)

    # Start time measurement
    start_time = time.time()

    # Sort the page weights using heap sort
    heapq.heapify(page_weights_list)
    sorted_weights = [heapq.heappop(page_weights_list) for _ in range(N)]
    sorted_weights.reverse()  # Get the weights in descending order

    # Extract the top K elements
    top_k_weights = sorted_weights[:K]

    # End time measurement
    end_time = time.time()
    elapsed_time = end_time - start_time

    # Output the results
    print("Top 20 weights:")
    for weight in top_k_weights:
        print(weight)
    print(f"Elapsed time: {elapsed_time:.6f} seconds")

if __name__ == "__main__":
    main()
```
